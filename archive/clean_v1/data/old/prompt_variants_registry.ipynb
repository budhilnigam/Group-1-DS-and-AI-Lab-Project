# Prompt Variant Registry (Simple Usage)

This notebook stores and uses multiple prompt variants from a single Python module.

Workflow:
1. Define prompt variants in src/rag_model/prompt_variants.py
2. Pick a variant name
3. Render a final prompt for any instance

```python
from pathlib import Path
import sys

ROOT = Path('..').resolve()
if str(ROOT) not in sys.path:
    sys.path.append(str(ROOT))

print('Project root:', ROOT)
```

```python
from src.rag_model.prompt_variants import (
    DEFAULT_PROMPT_VARIANT,
    list_prompt_variants,
    render_prompt,
)

print('Available variants:')
for name in list_prompt_variants():
    print('-', name)

print('Default variant:', DEFAULT_PROMPT_VARIANT)
```

```python
# Example input (replace with real data from your pipeline)
instance = {
    'repo': 'fastapi/fastapi',
    'pr_id': 'PR_1234',
    'file_path': 'fastapi/openapi/utils.py',
    'line_number': 396,
    'query_text': 'def fn(settings={\"mode\": \"fast\"}):\n    pass',
}

retrieved = [
    {
        'chunk_id': 'chunk_0737',
        'category': 'mutable_default',
        'source_type': 'augmented',
        'score': 0.6924,
        'text': "Default argument 'settings={...}' is mutable; use None and create inside function."
    },
]

predicted_category_hint = 'mutable_default'
```

```python
# Choose the prompt variant you want
selected_variant = DEFAULT_PROMPT_VARIANT
# Example: selected_variant = 'concise_review_v1'

final_prompt = render_prompt(
    variant_name=selected_variant,
    instance=instance,
    retrieved=retrieved,
    predicted_category_hint=predicted_category_hint,
)

print('Using variant:', selected_variant)
print('--- Prompt Preview ---')
print(final_prompt[:3000])
```
