# Synthetic Repos &amp; Review Comments

Generate synthetic GitHub repos with intentional violations, collect PR review comments, and merge them into the retrieval corpus.

```python
import subprocess, sys, os, json
from pathlib import Path
from dotenv import load_dotenv

load_dotenv(os.path.join(os.path.dirname(os.path.abspath("__file__")), "..", ".env"))

REPO_NAMES = ["flask", "fastapi", "pandas", "sklearn", "django"]
NUM_PRS = 20

REPO_TOKENS = [os.environ.get("GITHUB_REPO", "")]
LLM_TOKENS = [
    t.strip()
    for t in os.environ.get("GITHUB_LLM_TOKEN", "").strip("[]").split(",")
    if t.strip()
]

GUIDELINES_DIR = "data/raw/guidelines_raw"
REVIEW_OUTPUT = "data/raw/review_comments/review.json"
START_CHUNK = 217

PROJECT_ROOT = os.path.normpath(os.path.join(os.path.dirname(os.path.abspath("__file__")), ".."))
SCRIPT_DIR = os.path.join(PROJECT_ROOT, "scripts")

print(f"Repos       : {REPO_NAMES}")
print(f"PRs / repo  : {NUM_PRS}")
print(f"Repo tokens : {len(REPO_TOKENS)} token(s)")
print(f"LLM tokens  : {len(LLM_TOKENS)} token(s)")
print(f"Project root: {PROJECT_ROOT}")
```

```python
print("Step 1: Creating synthetic repos with violations & review comments\n")

result = subprocess.run(
    [
        sys.executable, "-u", os.path.join(SCRIPT_DIR, "create_synthetic_repos.py"),
        "--repo-token", REPO_TOKENS[0],
        "--llm-token", ",".join(LLM_TOKENS),
        "--repos", *REPO_NAMES,
        "--num-prs", str(NUM_PRS),
        "--guidelines-dir", GUIDELINES_DIR,
    ],
    cwd=PROJECT_ROOT, text=True,
)

print(f"\n{'[OK]' if result.returncode == 0 else f'[WARNING] exit code {result.returncode}'}")
```

```python
print("Step 2: Fetching PR review comments\n")

env = os.environ.copy()
env["GITHUB_REPO"] = REPO_TOKENS[0]

result = subprocess.run(
    [
        sys.executable, "-u", os.path.join(SCRIPT_DIR, "fetch_review_comments.py"),
        "--repos", *REPO_NAMES,
        "--start-chunk", str(START_CHUNK),
        "--output", REVIEW_OUTPUT,
    ],
    cwd=PROJECT_ROOT, env=env, text=True,
)

print(f"\n{'[OK]' if result.returncode == 0 else f'[WARNING] exit code {result.returncode}'}")
```

```python
print("Step 3: Merging review comments into retrieval corpus\n")

corpus_path = Path(PROJECT_ROOT) / "data" / "processed" / "retrival_corpus.json"
review_path = Path(PROJECT_ROOT) / REVIEW_OUTPUT

with open(corpus_path, "r", encoding="utf-8") as f:
    corpus = json.load(f)

with open(review_path, "r", encoding="utf-8") as f:
    reviews = json.load(f)

existing_ids = {entry["chunk_id"] for entry in corpus}
new_reviews = [r for r in reviews if r["chunk_id"] not in existing_ids]
merged = corpus + new_reviews

with open(corpus_path, "w", encoding="utf-8") as f:
    json.dump(merged, f, indent=2, ensure_ascii=False)

print(f"Existing corpus : {len(corpus)} chunks")
print(f"Review comments : {len(reviews)} chunks ({len(reviews) - len(new_reviews)} duplicates skipped)")
print(f"Newly added     : {len(new_reviews)} chunks")
print(f"Merged corpus   : {len(merged)} chunks")
print(f"Saved to {corpus_path}")
```

```python
print("Step 4: Merged Retrieval Corpus Analysis\n")

import pandas as pd
import matplotlib.pyplot as plt

with open(corpus_path, "r", encoding="utf-8") as f:
    merged_corpus = json.load(f)

df_merged = pd.DataFrame(merged_corpus)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

source_counts = df_merged["source_type"].value_counts()
source_counts.plot.bar(ax=axes[0], color="steelblue", edgecolor="black")
axes[0].set_title("Merged Corpus - Chunks per Source Type")
axes[0].set_ylabel("Count")
axes[0].set_xlabel("")
for i, v in enumerate(source_counts):
    axes[0].text(i, v + 0.5, str(v), ha="center", fontweight="bold")

cat_counts = df_merged["category"].value_counts()
cat_counts.plot.bar(ax=axes[1], color="coral", edgecolor="black")
axes[1].set_title("Merged Corpus - Chunks per Category")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("")
for i, v in enumerate(cat_counts):
    axes[1].text(i, v + 0.5, str(v), ha="center", fontweight="bold")

plt.tight_layout()
plt.show()

pivot = df_merged.groupby(["source_type", "category"]).size().unstack(fill_value=0)
ax = pivot.plot.bar(stacked=True, figsize=(14, 6), edgecolor="black", colormap="Set2")
ax.set_title("Merged Corpus - Category Breakdown per Source Type")
ax.set_ylabel("Count")
ax.set_xlabel("")
ax.legend(title="Category", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

print(f"Total chunks: {len(df_merged)} (guidelines: {len(corpus)}, reviews: {len(new_reviews)})\n")
print(pivot.to_string())
```

# Evaluation Dataset 


```python
EVAL_NUM_PRS = 20
EVAL_OUTPUT = "data/processed/evaluation.json"
EVAL_FILES_DIR = os.path.join(PROJECT_ROOT, "data", "processed", "evaluation_files")

print(f"Eval PRs/repo : {EVAL_NUM_PRS}")
print(f"Eval output   : {EVAL_OUTPUT}")
print(f"Eval files dir: {EVAL_FILES_DIR}")
print(f"Repos         : {REPO_NAMES}")
print(f"Total PRs     : {EVAL_NUM_PRS * len(REPO_NAMES)}")
```

```python
print("Step 5: Creating evaluation dataset from existing eval PRs\n")

cmd = [
    sys.executable, "-u", os.path.join(SCRIPT_DIR, "create_evaluation_dataset.py"),
    "--repo-token", REPO_TOKENS[0],
    "--llm-token", ",".join(LLM_TOKENS),
    "--repos", *REPO_NAMES,
    "--num-prs", str(EVAL_NUM_PRS),
    "--output", EVAL_OUTPUT,
]

result = subprocess.run(cmd, cwd=PROJECT_ROOT, capture_output=True, text=True, timeout=600)
print(result.stdout[-3000:] if len(result.stdout) > 3000 else result.stdout)
if result.stderr:
    for line in result.stderr.splitlines():
        if "Warning" not in line:
            print(f"STDERR: {line}")
print(f"\n{'[OK]' if result.returncode == 0 else f'[FAILED] exit code {result.returncode}'}")
```

```python
print("Step 5b: Merging manual framework-specific eval data into evaluation.json\n")

import json, shutil
from pathlib import Path

MANUAL_DIR = Path(PROJECT_ROOT) / "data" / "raw" / "eval_manual"
MANUAL_EVAL = MANUAL_DIR / "evaluation.json"
MANUAL_FILES = MANUAL_DIR / "eval_files"

generated_path = Path(PROJECT_ROOT) / EVAL_OUTPUT
generated_files_dir = Path(PROJECT_ROOT) / "data" / "processed" / "evaluation_files"

# Load generated eval data
with open(generated_path, "r", encoding="utf-8") as f:
    generated_data = json.load(f)

existing_ids = {e["id"] for e in generated_data}

# Load manual eval data
with open(MANUAL_EVAL, "r", encoding="utf-8") as f:
    manual_data = json.load(f)

# Copy manual source files and fix source_file paths
added_entries = []
added_files = 0
for entry in manual_data:
    if entry["id"] in existing_ids:
        print(f"  SKIP (duplicate id): {entry['id']}")
        continue

    # Copy .py file into evaluation_files/
    src_file = MANUAL_FILES / Path(entry["source_file"]).name
    dst_file = generated_files_dir / Path(entry["source_file"]).name
    if src_file.exists():
        shutil.copy2(src_file, dst_file)
        added_files += 1

    # Rewrite source_file to point to evaluation_files/
    entry["source_file"] = f"evaluation_files/{Path(entry['source_file']).name}"
    added_entries.append(entry)

# Merge and save
merged = generated_data + added_entries
with open(generated_path, "w", encoding="utf-8") as f:
    json.dump(merged, f, indent=2, ensure_ascii=False)

manual_reviews = sum(len(e["ground_truth_reviews"]) for e in added_entries)
gen_reviews = sum(len(e["ground_truth_reviews"]) for e in generated_data)

print(f"Generated entries   : {len(generated_data)} ({gen_reviews} reviews)")
print(f"Manual entries added: {len(added_entries)} ({manual_reviews} reviews)")
print(f"Manual files copied : {added_files}")
print(f"Merged total        : {len(merged)} entries ({gen_reviews + manual_reviews} reviews)")
print(f"Saved to {generated_path}")
```

```python
print("Step 6: Evaluation Dataset Analysis\n")

eval_path = Path(PROJECT_ROOT) / EVAL_OUTPUT

with open(eval_path, "r", encoding="utf-8") as f:
    eval_data = json.load(f)

print(f"Total evaluation entries: {len(eval_data)}")

total_reviews = sum(len(e["ground_truth_reviews"]) for e in eval_data)
print(f"Total ground_truth_reviews: {total_reviews}")

# Source files on disk
eval_files = list(Path(EVAL_FILES_DIR).glob("*.py")) if os.path.isdir(EVAL_FILES_DIR) else []
print(f"Source files in evaluation_files/: {len(eval_files)}")

# Line number stats
all_lines = [r["line_number"] for e in eval_data for r in e["ground_truth_reviews"]]
at_line1 = sum(1 for l in all_lines if l == 1)
print(f"Reviews at actual lines: {len(all_lines) - at_line1}, at line 1: {at_line1}")

import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

repo_counts = Counter(e["repo"] for e in eval_data)
cat_counts = Counter()
for e in eval_data:
    for r in e["ground_truth_reviews"]:
        cat_counts[r["violation_category"]] += 1

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

repos_s = pd.Series(repo_counts).sort_index()
repos_s.plot.bar(ax=axes[0], color="mediumseagreen", edgecolor="black")
axes[0].set_title("Evaluation - Entries per Repo")
axes[0].set_ylabel("Count")
axes[0].set_xlabel("")
for i, v in enumerate(repos_s):
    axes[0].text(i, v + 0.3, str(v), ha="center", fontweight="bold")

cats_s = pd.Series(cat_counts).sort_index()
cats_s.plot.bar(ax=axes[1], color="mediumpurple", edgecolor="black")
axes[1].set_title("Evaluation - Reviews by Violation Category")
axes[1].set_ylabel("Count")
axes[1].set_xlabel("")
for i, v in enumerate(cats_s):
    axes[1].text(i, v + 0.3, str(v), ha="center", fontweight="bold")

plt.tight_layout()
plt.show()

print(f"\nBy repo:")
for k, v in sorted(repo_counts.items()):
    print(f"  {k}: {v}")
print(f"\nBy violation category:")
for k, v in sorted(cat_counts.items()):
    print(f"  {k}: {v}")
```
